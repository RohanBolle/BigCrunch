```python
import sympy as sym
from sympy import symbols, Function, integrate, diff
from sympy.printing.latex import latex

#G_star

M, r = symbols('M r')
G = Function('G')(r)

# Define u and dv for integration by parts
# function = (M*G/r**2)

u = G
dv = M/r**2

# Calculate v
v = integrate(dv, r)

# Calculate du/dr
du = diff(u, r)

# Apply integration by parts formula
result = (u*v - integrate(v*diff(u, r), r))

print("Result:")
print(result)

# Convert to latex code
latex_result = latex(result)
print("\nLaTeX code for the equation:")
print(latex_result)
```

```python
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt
from sympy.printing import latex

# Define symbols
M, r, a, B, r_0, G_0 = sym.symbols('M r a B r_0 G_0')

# Define G function
G = G_0 * r**B

# Define functions
functions = {
    "G_star": G - r * sym.integrate(sym.diff(G, r) / r, r),
    "G1_star": (G**2) - (r**2) * sym.integrate(2*G * sym.diff(G, r) / (r**2), r),
    "G2_star": (G**2) - (2 * r**3) * sym.integrate(G * sym.diff(G, r) / (r**3), r)
}
for key, expr in functions.items():
    latex_result = latex(expr)  # Convert the expression to LaTeX
    print(f"\nLaTeX code for {key}:")  # Print the name of the equation
    print(latex_result)  # Print the LaTeX code

# Convert symbolic functions to numerical functions
lambdified_funcs = {key: sym.lambdify((r, G_0, B), expr, 'numpy') for key, expr in functions.items()}

# Define parameters
M_val = 1
r_0_val = 1
G_0_val = 1
r_vals = np.linspace(0.1, 1, 500)
B_vals = [1]

# Plot all functions
plt.figure(figsize=(8, 6))

for key, func in lambdified_funcs.items():
  for B_val in B_vals:
        result_vals = func(r_vals, G_0_val, B_val)
        plt.plot(r_vals, result_vals, label=f'{key}, B={B_val:.2f}')

plt.xlabel('r')
plt.ylabel('Result')
plt.title('$G_{0} a^{B}$')
plt.legend()
plt.grid()
# plt.yscale('log')
plt.savefig("Plot 1")
plt.show()
```

4) \begin{equation} G_{0}(1 + αr^n) \end{equation}

```python
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt

# Define symbols
M, r, alpha, n, r_0, G_0 = sym.symbols('M r alpha n r_0 G_0')

# Define G function
G = G_0*(1+alpha*(r**n))

# Define functions
functions = {
    "G_star": G - r * sym.integrate(sym.diff(G, r) / r, r),
    "G1_star": (G**2) - (r**2) * sym.integrate(G * sym.diff(G, r) / (r**2), r),
    "G2_star": (G**2) - (2 * r**3) * sym.integrate(G * sym.diff(G, r) / (r**3), r)
}

# Convert symbolic functions to numerical functions
lambdified_funcs = {key: sym.lambdify((r, G_0, alpha, n), expr, 'numpy') for key, expr in functions.items()}

# Define parameters
M_val = 1
r_0_val = 1
G_0_val = 1
r_vals = np.linspace(0.1, 1, 500)
alpha_vals = 1
n = 1

# Plot all functions
plt.figure(figsize=(8, 6))

for key, func in lambdified_funcs.items():
  result_vals = func(r_vals, G_0_val, alpha_vals, n)
  plt.plot(r_vals, result_vals, label=f'{key}, \u03B1 ={alpha_vals:.2f}')

plt.xlabel('r')
plt.ylabel('Result')
plt.title('$G_{0}(1 + αr^n)$')
plt.legend()
plt.grid()
plt.savefig("Plot 4")
plt.show()

for key, expr in functions.items():
    latex_result = latex(expr)
    print(f"\nLaTeX code for {key}:")
    print(latex_result)
```

G1_star \begin{equation}- G_{0}^{2} a n r^{2} \left(\begin{cases} a \log{\left(r \right)} - \frac{1}{r} & \text{for}\: n = 1 \\\frac{a r^{2}}{2} + \log{\left(r \right)} & \text{for}\: n = 2 \\\frac{a n r^{2 n}}{2 n^{2} r^{2} - 6 n r^{2} + 4 r^{2}} - \frac{2 a r^{2 n}}{2 n^{2} r^{2} - 6 n r^{2} + 4 r^{2}} + \frac{2 n r^{n}}{2 n^{2} r^{2} - 6 n r^{2} + 4 r^{2}} - \frac{2 r^{n}}{2 n^{2} r^{2} - 6 n r^{2} + 4 r^{2}} & \text{otherwise} \end{cases}\right) + G_{0}^{2} \left(a r^{n} + 1\right)^{2} \end{equation}

G2_star
\begin{equation}- 2 G_{0}^{2} a n r^{3} \left(\begin{cases} a \log{\left(r \right)} - \frac{2}{3 r^{\frac{3}{2}}} & \text{for}\: n = \frac{3}{2} \\\frac{a r^{3}}{3} + \log{\left(r \right)} & \text{for}\: n = 3 \\\frac{a n r^{2 n}}{2 n^{2} r^{3} - 9 n r^{3} + 9 r^{3}} - \frac{3 a r^{2 n}}{2 n^{2} r^{3} - 9 n r^{3} + 9 r^{3}} + \frac{2 n r^{n}}{2 n^{2} r^{3} - 9 n r^{3} + 9 r^{3}} - \frac{3 r^{n}}{2 n^{2} r^{3} - 9 n r^{3} + 9 r^{3}} & \text{otherwise} \end{cases}\right) + G_{0}^{2} \left(a r^{n} + 1\right)^{2}\end{equation}

```python
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import expi
from sympy.printing.latex import latex


M, r, alpha, n, r_0, G_0 = sym.symbols('M r alpha n r_0 G_0')
G = G_0*(1+alpha*(r**n))

func = sym.diff(G, r)/r

# Define u and dv
result = sym.integrate(func, r)
ans = G-r*result
print("Result:")
sym.pprint(ans)

# Convert to latex code
latex_result = latex(ans)
print("\nLaTeX code for the equation:")
print(latex_result)


M_val = 1
r_0_val = 1
G_0_val = 1
alpha_val = 1
n_values = [0.5, 1]
r_vals = np.linspace(0.1, 10, 500)

for n_val in n_values:
    result_vals = [ans.subs({r: rv, M: M_val, r_0: r_0_val, G_0: G_0_val, alpha: alpha_val, n: n_val}).evalf() for rv in r_vals]
    plt.plot(r_vals, result_vals, label=f"n = {n_val}")

# result_vals_1 = [ans.subs({r: rv, M: M_val, r_0: r_0_val, G_0: G_0_val, alpha: alpha_val, n:n_val}).evalf() for rv in r_vals]

plt.xlabel('r')
plt.ylabel('Result')
plt.title('G0_star')
plt.legend()
plt.grid()
plt.show()
```
